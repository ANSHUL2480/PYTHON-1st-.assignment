2.)Describe the role of predefined keywords in Python and provide examples of how they are used in a
program.

Ans.)In Python, predefined keywords (also called reserved keywords) are words that have special meaning in the language's syntax. These keywords are reserved for specific programming constructs and cannot be used as variable names or identifiers.

 Role of Predefined Keywords in Python:
1.Control Flow: Keywords like if, else, elif, for, while, and break control the logic and flow of the program.

2.Function and Class Definitions: Keywords like def and class are used to define functions and classes.

3.Data Handling: Keywords like in, is, not, and and help in handling data logic and conditions.

4.Error Handling: Keywords like try, except, finally, and raise are used to handle exceptions.

5.Variable Scope: Keywords like global, nonlocal control variable scope.

Others: Keywords like import, return, with, yield play specialized roles.

Example in program


```python
# Using various predefined keywords

def greet(name):  # 'def' keyword to define a function
    if name:  # 'if' keyword to check condition
        print(f"Hello, {name}!")  # 'print' is a built-in function
    else:  # 'else' keyword for alternative path
        print("Hello, stranger!")

# Looping with 'for'
names = ["Alice", "Bob", "Charlie"]
for person in names:  # 'for' and 'in' used for looping
    greet(person)

# Handling exceptions
try:  # 'try' block to catch errors
    x = 10 / 0
except ZeroDivisionError:  # 'except' to handle specific exception
    print("Cannot divide by zero.")
finally:  # 'finally' always runs
    print("Done with division operation.")
```

3.) Compare and contrast mutable and immutable objects in Python with examples.

Ans.)In Python, objects are broadly classified as mutable or immutable based on whether their state (or content) can be changed after they are created.

🔁 Mutable vs ❄️ Immutable Objects in Python
Feature	Mutable Objects	Immutable Objects
Can be changed after creation?	✅ Yes	❌ No
Examples	list, dict, set, bytearray	int, float, str, tuple, frozenset, bytes
Memory Address (id)	Can remain the same even after changes	New object (new id) is created on change
Usage	When you need to modify data in place	When you need hashable (constant) values

✅ Mutable Object Example: list


```python
my_list = [1, 2, 3]
print(id(my_list))   # e.g., 140000000000
my_list.append(4)
print(my_list)       # [1, 2, 3, 4]
print(id(my_list))   # Same as before, object changed in place
```

❌ Immutable Object Example: str

```python
my_str = "hello"
print(id(my_str))    # e.g., 140000000111
my_str += " world"
print(my_str)        # "hello world"
print(id(my_str))    # Different from before; new object created
```

```python
[28]
0s
x = 10
if x > 5:
print("x is greater than 5")
x is greater than 5
```

Provides an alternative block of code to run if the condition is false.

```python
[29]
0s
x = 3
if x > 5:
print("x is greater than 5")
else:
print("x is 5 or less")
x is 5 or less
```
